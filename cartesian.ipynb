```python
import math

import numpy as np
np.set_printoptions(suppress=True)
```

```python
def cts(q):
    x = q[0]
    y = q[1]
    z = q[2]
    rho = np.sqrt(x**2+y**2+z**2)
    theta = float(math.degrees(np.arccos(z/rho)))
    phi = np.sign(y)*float(math.degrees(np.arccos(x/np.sqrt(x**2+y**2))))
    q_s = np.array([rho,theta,phi])
    return q_s
def stc(q):
    rho = (q[0])
    theta = math.radians(q[1])
    phi= math.radians(q[2])
    x = rho*np.sin(theta)*np.cos(phi)
    y = rho*np.sin(theta)*np.sin(phi)
    z = rho*np.cos(theta)
    q_s = np.array([x,y,z])
    return q_s
```
