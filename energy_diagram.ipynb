```python
import matplotlib.pyplot as plt
plt.style.use("default")

def plot_energy_diagram(positions, energies, labels, connection, x_label= False, title="Molecular energy diagram", save = False, fig_shape = [10,6]):
    """
    states  : list of str, names/labels of the states (e.g. ['S0', 'S1', 'T1'])
    energies: list of float, energies in eV, kJ/mol, etc. (same units)
    """

    if len(positions) != len(energies) != len(labels):
        raise ValueError("states and energies must have the same length")
    levels = zip(energies, labels)

    fig, ax = plt.subplots(figsize=(fig_shape[0], fig_shape[1]))

    for x, (E, label) in zip(positions, levels):
        # Draw a horizontal line for the energy level
        ax.hlines(E, x - 0.3, x + 0.3, color="red", linewidth=2)
        # Label next to the line
        ax.text(x, E, f"{label}\n({E:.1f})",
            va="center", ha="center", fontsize=9)

    # Formatting
    ax.set_xticks([])
    ax.set_ylabel("Energy (kcal/mol)")
    ax.set_title(title)
    ymax = max(energies) + 0.1 * abs(max(energies))
    # Make the y-limits a bit padded
    if min(energies) == 0:
        ymin = - 0.1 * abs(max(energies))
    else:
        ymin = min(energies) - 0.1 * abs(min(energies))

    ax.set_ylim(ymin, ymax)

    plt.tight_layout()
    dashed_line(positions[0:2], energies[0:2])
    for i, el in enumerate(connection):
        for j, el2 in enumerate(connection[i+1:], start=i+1):
            if el == el2 and position[i] == position[j] -1:
                dashed_line([positions[i], positions[j]], [energies[i], energies[j]])

    if x_label != False:
        plt.xticks(range(len(x_label)), x_label)

    if save:
        plt.savefig(f"{title.replace(" ","_")}.png", dpi=300, bbox_inches="tight")

    plt.show()
def dashed_line(xs, ys):
    x_dash = [xs[0] + 0.3 ,xs[1] - 0.3]
    y_dash = [ys[0], ys[1]]
    plt.plot(x_dash, y_dash, color="red", linestyle="dashed")
    return
```

```python
connection = [0, 0, 0, 1, 1, 1]
labels = ["Whole complex", "COax no opt", "Square Piramidal Cl ax", "Whole complex", "COeq no opt", "Square Piramidal Cl eq"]
energies = [0, 74.95, 71.45, 0, 68.42, 51.54]
position = [0, 1, 2, 0, 1, 2]
plot_energy_diagram(position, energies, labels, connection, title="S0 Vacuum facial CO ligands dissociation energy diagram", save = True)
```
