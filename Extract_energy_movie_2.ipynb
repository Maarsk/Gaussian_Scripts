```python
import csv
import re
import glob
import os
```

```python
def make_folder(input_f,output):
    os.makedirs(output, exist_ok=True)
    os.makedirs(input_f, exist_ok=True)
```

```python
def extract_energies(xyz_file):
    energies = []

    # Matches: Energy: -1373.2881252
    energy_pattern = re.compile(r"Energy:\s*([-+]?\d*\.\d+(?:[eE][-+]?\d+)?)")

    with open(xyz_file, "r") as f:
        for line in f:
            match = energy_pattern.search(line)
            if match:
                energies.append(float(match.group(1)))

    # Output CSV path
    csv_file = xyz_file.split(".")[0] + "_MovieEnergies.csv"
    csv_file = csv_file.replace("Input\\", "Output\\")

    # Write CSV
    with open(csv_file, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["Energy"])
        for e in energies:
            writer.writerow([e])

    print(f"Extracted {len(energies)} energies → saved to {csv_file}")
```

```python
input_folder = "Input\\movie"
output_folder = "Output\\movie"
make_folder(input_folder,output_folder)
for xyz_file in glob.glob(input_folder+"/*.xyz"):
    extract_energies(xyz_file)
```
